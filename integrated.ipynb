```python
import pandas as pd
df = pd.read_csv('for_export_philippine_elections.csv')
df.head()

authors = pd.read_csv('well_known_authors_philippine_elections.csv')
```

```python
import re

def filter_text(text):
    new_text = re.sub(r'https?://\S+|www\.\S+', r'', str(text))
    new_text = re.sub(r'\@\S+', r'', str(new_text))
    new_text = re.sub(r'[^\x00-\x7f]', r'', str(new_text))
    new_text = re.sub(r'[^\w\s\@\#]', r' ', str(new_text))
    new_text = new_text.lower()
    return new_text

df['text_filtered'] = df['text'].apply(filter_text)

df.head()
```

```python
df['pseudo_author_userName'] = df['pseudo_author_userName'].astype('Int64').astype('string')
df['pseudo_inReplyToUsername'] = df['pseudo_inReplyToUsername'].astype('Int64').astype('string')

author_usernames = dict(zip(
    authors['obfuscated_userName'].astype(str).str[1:], 
    authors['author_userName']
))
print(author_usernames)

for i in range(len(df)):
    id = df.iloc[i]['pseudo_author_userName']
    if id in author_usernames:
        df.at[i, 'pseudo_author_userName'] = author_usernames[id]
    if pd.notna(df.iloc[i]['pseudo_inReplyToUsername']):
        id2 = df.iloc[i]['pseudo_inReplyToUsername']
        if id2 in author_usernames:
            df.at[i, 'pseudo_inReplyToUsername'] = author_usernames[id2] 
```

```python
import stopwordsiso as stopwords

grouped_df = []

for i in range(len(communities)):
    grouped_df.append(df[df['pseudo_author_userName'].isin(str(x) for x in communities[i])])

word_tokens_list = []

for i in range(len(communities)):
    word_tokens_list.append([tweet.split() for tweet in grouped_df[i]['text_filtered']])

# stop words for english and tagalog
stop_words = stopwords.stopwords(["en", "tl"])
stop_words.update(['yan', 'yang', 'naman', 'lang', 'yung', 'di', 'niyo', 'nyo', 'sana', 'na', 'nga', 'kasi', 'po', 'natin', 'mag',
                    'kay', 'talaga', 'nalang', 'jan', 'nung', 'parang', 'pala', 'ung', 'kayong', 'mas', 'rin', 'din', 'kayo',
                   'pag', 'rin', 'daw', 'raw', 'yun', 'nag', 'sya', 'nya', 'nang', 'upang', 'ayon', 'upang', 'nitong', 'pang', 
                   'inyo', 'dyan', 'lng', 'ngayong', 'wag', 'baka', 'pls', 'para', 'nasa', 'muna', 'sige', 'puro', 'lalo', 'mong',
                   'ibang', 'anong'])
stop_words.update(['binoto', 'botohan', 'boto', 'vote', 'iboto', 'halalan', 'iboboto', 'bumoto', '#halalan2025', 'eleksyon2025',
                   'iboboto', 'boto', 'vote', 'votes', 'elections', 'election', 'politics', '2025', 'philippine', 'voting',
                   'pilipinas', 'kc', 'nating', 'ninyo', 'nmn', 'ha', 'haha', 'hahaha', 'eme', 'eto', 'iyan', 'maka', 'tong',
                   'kang', 'itong', 'boboto', 'ilang', 'oo', 'pinag', 'yon', 'halalan2025', '1', '2', '3', 'nman', 'lol', 'taga',
                   'kita', 'yo', 'naku', ''])
# for removing fandom related voting events
stop_words.update(['sb19', 'stell', 'jhoaiah', 'blooms', 'colaiah', 'barriofiestadxbjmfyang', 'jmfyang', 'bini', 'bebeBerries', 'jl',
                   'congrats', 'dustin', '#savewithmaya', 'housemate'])
filtered_tokens = []
for i in range(len(communities)):
    filtered_tokens.append([])
    for word_tokens in word_tokens_list[i]:
        filtered_tokens[i].append([token for token in word_tokens if (token.lower() not in stop_words)])


# for x in temp_filtered_tokens:
#     print(x)
# print("------------------")
for x in filtered_tokens:
    print("----------------")
    for y in x:
        print(y)
```

```python
from gensim.corpora.dictionary import Dictionary
from gensim.models.ldamodel import LdaModel

dictionary = []
corpus = []
lda = []

for i in range(len(communities)):

    dictionary.append(Dictionary(filtered_tokens[i]))
    corpus.append([dictionary[i].doc2bow(text) for text in filtered_tokens[i]])

    lda.append(LdaModel(
        corpus=corpus[i], 
        id2word=dictionary[i], 
        num_topics=4
    ))
```

```python
import pyLDAvis, pyLDAvis.gensim_models

pyLDAvis.enable_notebook()

for i in range(len(communities)):
    vis = pyLDAvis.gensim_models.prepare(lda[i], corpus[i], dictionary[i])
    pyLDAvis.save_html(vis, "integrated/results" + str(i) + ".html")
```
